Baixar o dataset

Ler um dos arquivos CSV (CAvideos.csv) do dataset

```python
csv_file = os.path.join(path, "CAvideos.csv")

dados = pd.read_csv(csv_file)
```

Faz uma cópia com apenas algumas colunas

```python
dados_novos = dados[['category_id', 'title', 'channel_title',
                     'views', 'likes', 'dislikes',
                     'comment_count']].copy()
```

Visualizar o dataset

```python
dados_novos
```

Converter para float

```python
pd.options.display.float_format = lambda x : f'{x:,.2f}'.replace(",", "X").replace(".", ",").replace("X", ".")

dados_novos['views'] = dados_novos['views'].astype(float)
dados_novos['likes'] = dados_novos['likes'].astype(float)
dados_novos['dislikes'] = dados_novos['dislikes'].astype(float)
dados_novos['comment_count'] = dados_novos['comment_count'].astype(float)
```

Ordenar pela coluna views em ordem decrescente

```python
dados_novos.sort_values(by='views', ascending=False)
```

Importar biblioteca math para usar funções como log10

```python
import math
```

Criando a estrutura de abreviaturas para Mil, Milhões (MI), Bilhões (BI) e Trilhões (TRI)

```python
numerosAbreviados = pd.DataFrame({'abreviaturas' : [
    'mil', 'MI', 'BI', 'TRI'
] })

def numeroAbr(x):
  expoente = int(math.log10(x) / 3)

  if(expoente >= 1):
    numero = x / 10**(expoente * 3)
    numero = f'{numero:,.2f} {numerosAbreviados.iloc[expoente-1]['abreviaturas']}'
    return numero
  return x
```

Apenas testar a saída da função criada: numeroAbr

```python
numeroAbr(9990)
```

Inserir ou atualizar novas colunas do dataset

```python
if 'view abr' not in dados_novos.columns:
  dados_novos.insert(4, 'view abr', dados_novos['views']
                   .map(numeroAbr))
else:
  dados_novos['view abr'] = dados_novos['views'].map(numeroAbr)
```

```python
import os
import pandas as pd
import kagglehub as kh
```

```python
path = kh.dataset_download('datasnaek/youtube-new')
```

```python
pathJoin = os.path.join(path, 'CAvideos.csv')
dados = pd.read_csv(pathJoin)
```

```python
#Exemplo para apagar colunas
dados.drop(columns=['trending_date', 'thumbnail_link', 'ratings_disabled'], inplace=True, errors='ignore')
```

```python
dados_some_columns = dados[['category_id', 'video_id',
                            'title', 'views', 'likes',
                            'dislikes']].copy()
```

```python
def relacaoMetricaXViews(metrica, views):
  return round((metrica / views) * 100, 2)
```

```python
qtdColunas = len(dados_some_columns.columns)
```

```python
dados_some_columns.columns
```

Adicionar a coluna RelacaoCurtidas

```python
#Verificar se já existe a coluna RelacaoCurtidas
if 'relacaoCurtidas' not in dados_some_columns.columns:
  #Inserir a coluna RelacaoCurtidas
  dados_some_columns.insert(5,
                            'relacaoCurtidas',
                            dados_some_columns.apply(
                                lambda linha :
                                relacaoMetricaXViews(linha['likes'], linha['views'])
                                , axis=1
                              ))


# Atualizar a coluna RelacaoCurtidas
dados_some_columns['relacaoCurtidas'] = dados_some_columns.apply(lambda linha : relacaoMetricaXViews(linha['likes'], linha['views']), axis=1)
```
